```python
import joblib
import pandas as pd
from scipy.stats import kurtosis, skew

# Load saved artifacts
model = joblib.load("model.pkl")
scaler = joblib.load("scaler.pkl")
final_columns = joblib.load("columns.pkl")

# Load the exact same file used in Streamlit
df_init = pd.read_csv("test_df.csv")
df_init=df_init.iloc[[x for x in range(10)]]

df = pd.DataFrame()
acc = ["a1", "a2", "a3", "a4"]
window = 10

for col in acc:
    df[f"{col}_std"] = df_init[col].groupby(df_init.index//window).std()

for col in acc:
    df[f"{col}_min"] = df_init[col].groupby(df_init.index//window).min()

for col in acc:
    df[f"{col}_max"] = df_init[col].groupby(df_init.index//window).max()

for col in acc:
    df[f"{col}_var"] = df_init[col].groupby(df_init.index//window).var()

for col in acc:
    df[f"{col}_range"] = df_init[col].groupby(df_init.index//window)\
                                     .agg(lambda x: x.max()-x.min())

for col in acc:
    df[f"{col}_kurtosis"] = df_init[col].groupby(df_init.index//window)\
                                        .agg(lambda x: kurtosis(x))

for col in acc:
    df[f"{col}_skew"] = df_init[col].groupby(df_init.index//window)\
                                    .agg(lambda x: skew(x))

df["load"] = df_init["load"].groupby(df_init.index//window)\
                            .agg(lambda x: x.mode().iloc[0])

df = df.fillna(0)
df = df.reindex(columns=final_columns, fill_value=0)

X = scaler.transform(df)

print(model.predict(X))
```
